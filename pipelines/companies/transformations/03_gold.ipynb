```python
import dlt
from pyspark.sql.functions import col, count

# Criação da tabela GOLD agregada
@dlt.table(
    name="lab.gold.companies",
    comment="Relatório GOLD com quantidade de empresas por porte"
)
def ldz_gold_relatorio_porte_empresa():

    df = spark.read.table(
        "lab.silver.companies"
    )

    return (
        df.groupBy("porte_empresa")
          .agg(
              count("cnpj_basico").alias("qtd_empresas")
          )
          .orderBy(col("qtd_empresas").desc())
    )
```
