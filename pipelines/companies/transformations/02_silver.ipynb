```python
import dlt
from pyspark.sql import functions as F

# Tabelas
BRONZE_TABLE = "lab.bronze.companies"
SILVER_TABLE = "lab.silver.companies"


@dlt.table(
    name=SILVER_TABLE,
    comment="Camada Silver com a última competência (ano_mes) da Receita Federal Empresas"
)
def receita_federal_empresas_silver():

    # Leitura da bronze
    df_bronze = spark.read.table(BRONZE_TABLE)

    # Recupera o maior ano_mes
    df_max = (
        df_bronze
        .agg(F.max("ano_mes").alias("ano_mes"))
    )

    # Filtra somente a última competência
    df_result = (
        df_bronze
        .join(df_max, on="ano_mes", how="inner")
    )

    return df_result
```
